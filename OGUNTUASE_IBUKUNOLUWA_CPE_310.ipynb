```python
class InsufficientFundsError(Exception):
    """Raised when there is not enough money in the account."""
    pass


class BankAccount:
    transaction_fee = 0.50

    def __init__(self, owner, matric_no, initial_balance=0.0):
        if initial_balance < 0:
            raise ValueError("Initial balance cannot be negative.")

        self.owner = owner
        self.matric_no = matric_no
        self._balance = float(initial_balance)

    @property
    def balance(self):
        return self._balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit amount must be positive.")

        self._balance += amount

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive.")

        total = amount + BankAccount.transaction_fee

        if total > self._balance:
            raise InsufficientFundsError("Insufficient funds.")

        self._balance -= total

    @classmethod
    def from_dict(cls, data):
        return cls(
            data["owner"],
            data["matric_no"],
            data["balance"]
        )

    def __str__(self):
        return (
            f"Owner Name : {self.owner}\n"
            f"Matric No  : {self.matric_no}\n"
            f"Balance    : ₦{self.balance:.2f}"
        )

    def __repr__(self):
        return (
            f"BankAccount('{self.owner}', "
            f"'{self.matric_no}', {self.balance})"
        )
```

```python

# Create account
account = BankAccount(
    "Oguntuase ibukunoluwa",
    "CPE/2021/1043",
    0
)

print("===== ACCOUNT DETAILS ====")
print(account)


# Deposit
print("\nDepositing ₦5000...")
account.deposit(5000)
print(account)


# Withdraw
print("\nWithdrawing ₦1000...")
account.withdraw(1000)
print(account)


# Test invalid deposit
try:
    account.deposit(-100)
except ValueError as e:
    print("\nDeposit Error:", e)


# Test insufficient funds
try:
    account.withdraw(10000)
except InsufficientFundsError as e:
    print("\nWithdrawal Error:", e)


# Create another account from dictionary
data = {
    "owner": "Oguntuase ibukunoluwa",
    "matric_no": "CPE/2021/1043",
    "balance": 0
}

new_account = BankAccount.from_dict(data)

print("\n===== ACCOUNT CREATED FROM DICTIONARY ====")
print(new_account)

print("\nrepr() output:")
print(repr(new_account))
```
